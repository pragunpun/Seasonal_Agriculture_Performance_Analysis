# Data Cleaning

```python
# Check data types

print(df.dtypes)
```

```python
# Check missing values

missing = df.isnull().sum()

missing[missing > 0]
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing.values,
    y=missing.index
)

plt.title('Missing Values by Column')
plt.xlabel('Number of Missing Values')
plt.ylabel('Column')

plt.tight_layout()
plt.show()
```

```python
# Fill missing values with median

df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())
```

Missing values in numerical variables were replaced using the median because median is less affected by outliers than the mean. Agricultural data such as rainfall, soil moisture and Yield can have extreme values, so median imputation is safer when the data is skewed.

# Outlier Analysis

Outliers are observations that are unusually high or low compared with the rest of the data.

An outlier should not automatically be deleted. In agriculture, extreme rainfall, yield, production, cost or profit values may represent genuine conditions. 

```python
# Boxplots for numerical variables

for column in numerical_columns:
    plt.figure(figsize=(8, 3))
    
    sns.boxplot(data=df, x=column)
    
    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    
    plt.show()
```

```python
# Outlier analysis using IQR method

outlier_summary = []

for column in numerical_columns:
    
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    
    outlier_summary.append({
        "Variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier %": (len(outliers) / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(outlier_summary)
```

```python
# Variables with the highest number of potential outliers

display(outlier_summary.sort_values("Outlier Count", ascending=False))
```

A correlation value ranges from -1 to +1:
- +1 -> Perfect positive relationship
- Close to +1 -> Strong positive relationship
- Around 0 -> Weak/no linear relationship
- Close to -1 -> Strong negative relationship
- -1 -> Perfect negative relationship
